```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

plt.rcParams["figure.dpi"] = 400
plt.style.use("seaborn-v0_8-pastel")
sns.set_palette([sns.color_palette()[i] for i in [1, 2, 0]])

data = pd.read_csv("linechart.csv")
```

```python
plt.rcParams["figure.figsize"] = (4, 3)

lineplot = sns.lineplot(data[data["Model"] == "VIT-B-16"], x="BS", y="Peak Memory Usage (MB)", hue="Framework", errorbar=("ci", False))

markers = ["v", "X", "o"]

for i, line in enumerate(lineplot.get_lines()):
    line.set_marker(markers[i % len(markers)])

plt.xlim([1, 36])
for x in plt.xticks()[0]:
    plt.axvline(x, color="black", alpha=0.05)
    
for y in plt.yticks()[0]:
    plt.axhline(y, color="black", alpha=0.05)
    
legend = plt.legend(loc="upper center", ncol=3, bbox_to_anchor=(0.4, 1.15))
plt.savefig("latency-b-16.png", bbox_inches="tight", dpi=400)
```
