```python
fig, axes = plt.subplots(4, 2, figsize=(14, 14))

curr = axes[0][0]
curr.set_title("ResNet (Loss)", fontsize=fontsize)

ax = sns.lineplot(loss, x="iteration", y="resnet", ax=curr, label="ResNet", color=colors[0], linestyle="--", marker="o")
sns.lineplot(loss, x="iteration", y="revnet", ax=curr, label="RevNet", color=colors[2], linestyle=":", marker="v")
sns.lineplot(loss, x="iteration", y="rev_resnet", ax=curr, label="Rev-ResNet", color="black")
curr.set_xlabel(None)
curr.set_ylabel(None)

for tick in ax.get_yticks():
    ax.axhline(y=tick, color="gray", linestyle="--", zorder=0, alpha=0.2)

curr = axes[0][1]
curr.set_title("ResNet (Acc)", fontsize=fontsize)

ax = sns.lineplot(acc, x="iteration", y="resnet", ax=curr, label="ResNet", color=colors[0], linestyle="--", marker="o")
sns.lineplot(acc, x="iteration", y="revnet", ax=curr, label="RevNet", color=colors[2], linestyle=":", marker="v")
sns.lineplot(acc, x="iteration", y="rev_resnet", ax=curr, label="Rev-ResNet", color="black")
curr.set_xlabel(None)
curr.set_ylabel(None)

for tick in ax.get_yticks():
    ax.axhline(y=tick, color="gray", linestyle="--", zorder=0, alpha=0.2)

def draw(row, model):
    left, right = row
    left.set_title(f"{model} (Loss)", fontsize=fontsize)
    right.set_title(f"{model} (Acc)", fontsize=fontsize)
    
    ax = sns.lineplot(loss, x="iteration", y=model.lower(), ax=left, label=model, color=colors[0], linestyle="--", marker="o")
    sns.lineplot(loss, x="iteration", y=f"rev_{model.lower()}", ax=left, label=f"Rev-{model}", color="black")
    left.set_xlabel(None)
    left.set_ylabel(None)
    
    for tick in ax.get_yticks():
        ax.axhline(y=tick, color="gray", linestyle="--", zorder=0, alpha=0.2)
    
    ax = sns.lineplot(acc, x="iteration", y=model.lower(), ax=right, label=model, color=colors[0], linestyle="--", marker="o")
    sns.lineplot(acc, x="iteration", y=f"rev_{model.lower()}", ax=right, label=f"Rev-{model}", color="black")
    right.set_xlabel(None)
    right.set_ylabel(None)
    
    for tick in ax.get_yticks():
        ax.axhline(y=tick, color="gray", linestyle="--", zorder=0, alpha=0.2)
    
draw(axes[1], "ViT")
draw(axes[2], "Swin")
draw(axes[3], "DeiT")

axes[-1][0].set_xlabel("Epoch", fontsize=fontsize)
axes[-1][1].set_xlabel("Epoch", fontsize=fontsize)

plt.savefig("loss.pdf", bbox_inches="tight", dpi=300)
```
